### Task 0

```python
%%sql
SELECT
    DATE(orders.orderDate, 'start of year') AS order_year, 
    customers.city                          AS city, 
    COUNT(*)                                AS cnt, 
    COUNT(DISTINCT orders.customerID)       AS distinct_customer 
FROM orders
    INNER JOIN customers
        ON customers.customerId = orders.customerId
WHERE orderDate >= '2016-01-01' AND orderDate < '2017-01-01' 
GROUP BY 1, 2
ORDER BY 3 DESC
LIMIT 5 
```

### Task 4

```python
%%sql
SELECT 
    CategoryID, 
    COUNT(ProductID) AS productCnt
FROM Products
GROUP BY CategoryID
ORDER BY CategoryID
```

### Task 5

```python
%%sql
SELECT 
    CategoryID, 
    ROUND(MIN(UnitPrice),2) AS minPrice,
    ROUND(MAX(UnitPrice),2) AS maxPrice,
    ROUND(AVG(UnitPrice),2) AS avgPrice
FROM Products
GROUP BY CategoryID
ORDER BY CategoryID
```

### Task 6

```python
%%sql
SELECT    
    DATE(OrderDate, 'start of month') AS monthStart,
    COUNT(OrderID) AS ordersCount 
FROM ORDERS
GROUP BY monthStart
ORDER BY monthStart
LIMIT 10
```

### Task 7

```python
%%sql
SELECT  
    DATE(OrderDate, 'start of year') AS yearStart,
    COUNT(OrderID) AS ordersCount   
FROM ORDERS
GROUP BY yearStart
ORDER BY yearStart
```

### Task 9

```python
%%sql
SELECT 
    o.OrderID,
    c.CompanyName,
    o.OrderDate,    
    c.CustomerID
FROM Orders AS o
JOIN Customers AS c   
ON o.CustomerID = c.CustomerID
ORDER BY o.CustomerID
LIMIT 10
```

### Task 10

```python
%%sql
SELECT 
    c.City,
    COUNT(DISTINCT c.CustomerID) AS CustomersCount,
    COUNT(o.OrderID) AS OrdersCount
FROM Orders AS o
JOIN Customers AS c   
ON o.CustomerID = c.CustomerID
GROUP BY c.City
ORDER BY CustomersCount DESC
LIMIT 10
```
