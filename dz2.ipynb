## 1

```python
%%sql
WITH neworders AS (
    SELECT 
        orderId, 
        customerId, 
        orderDate
    FROM 
        Orders
    WHERE 
        DATE(orders.orderDate, 'start of year') = '2012-01-01'
)
SELECT 
    orderId, 
    customerId, 
    orderDate
FROM 
    neworders
ORDER BY 
    orderDate
LIMIT 10
```

## 2

```python
%%sql
DROP TABLE IF EXISTS NPRODUCTS;
CREATE TABLE NPRODUCTS AS
SELECT
    productId,
    UnitPrice
FROM 
    Products
WHERE 
    UnitPrice > (SELECT AVG(UnitPrice) FROM Products);

SELECT 
    productId,
    UnitPrice
FROM 
    NPRODUCTS
ORDER BY 
    UnitPrice DESC 
LIMIT 10 
```

## 3

```python
%%sql
SELECT 
    -- OrderDate AS 2012,
    2012 AS year,
    COUNT(*) AS OrderCnt
FROM 
    Orders
WHERE 
    DATE(orders.orderDate, 'start of year') = '2012-01-01'
UNION ALL
SELECT 
    -- OrderDate AS 2013,
    2013 AS year,
    COUNT(*) AS OrderCnt
FROM 
    Orders
WHERE 
    DATE(orders.orderDate, 'start of year') = '2013-01-01'
```

## 4

```python
%%sql
SELECT
    CustomerID,
    CompanyName,
    'London' AS City
FROM Customers
WHERE city = 'London'
UNION ALL
SELECT
    CustomerID,
    CompanyName,
    'Paris' AS City
FROM Customers
WHERE city = 'Paris'
ORDER BY
    CustomerID
LIMIT 10
```

## 5

```python
%%sql
SELECT
    ProductID,
    ProductName,
    UnitPrice,
    CASE
        WHEN UnitPrice > 50 THEN 'Expensive'
        WHEN UnitPrice BETWEEN 20 AND 50 THEN 'Moderate'
        WHEN UnitPrice < 20 THEN 'Cheap'     
    END AS PriceCategory
FROM Products
ORDER BY ProductID
LIMIT 10
```

## 6

```python
%%sql
SELECT
    CASE
        WHEN OrderDate < '2014-01-01' THEN 'Old'
        ELSE 'New'
    END AS OrderStatus,
    COUNT(*) AS OrderCnt
FROM Orders
GROUP BY OrderStatus   
ORDER BY OrderStatus
```

## 7

```python
%%sql
WITH NumberedOrders AS (
    SELECT
        OrderId,
        ShipName,
        ShippedDate,
        ROW_NUMBER() OVER (PARTITION BY ShipName ORDER BY ShippedDate) AS rn
    FROM Orders
)
SELECT
    OrderId,
    ShipName,
    ShippedDate,
    rn
FROM NumberedOrders
WHERE ShipName = 'Ana Trujillo Emparedados y helados'
ORDER BY rn
LIMIT 10
```

## 8

```python
%%sql
SELECT 
    CustomerId, 
    OrderDate,
    COUNT(*) OVER(PARTITION BY customerId ORDER BY OrderDate) AS CumulativeOrders
FROM Orders
ORDER BY CustomerId DESC
LIMIT 10
```

## 9

```python
%%sql
SELECT 
    CustomerId,
    OrderID,
    OrderDate,
    LAG(OrderDate) OVER(PARTITION BY CustomerId ORDER BY OrderDate) AS PrevOrderDate,
    ROUND(julianday(OrderDate) - julianday(LAG(OrderDate) OVER(PARTITION BY CustomerId ORDER BY OrderDate))) AS DaysSincePrevOrder
FROM Orders
WHERE CustomerId = 'WILMK'
ORDER BY OrderId
LIMIT 10
```
